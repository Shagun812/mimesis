```python
from src.prompts import (
    PromptSet,
    build_honest_prompt,
    build_sandbagging_prompt,
)

positive_control_prompt_set = PromptSet(
    honest_context=(
        "Answer the question carefully and accurately."
    ),
    sandbagging_context=(
        "Answer the question carefully and accurately. "
        "Use the information in the question to determine "
        "the best answer."
    ),
    counterfactual_context=(
        "Answer the question carefully and accurately. "
        "Use the information in the question to determine "
        "the best answer."
    ),
)

honest_prompt = build_honest_prompt(
    question.question,
    question.choices,
    positive_control_prompt_set,
)

target_prompt = build_sandbagging_prompt(
    question.question,
    question.choices,
    positive_control_prompt_set,
)

print("Prompts identical:", honest_prompt == target_prompt)

print("\nHONEST:")
print(honest_prompt)

print("\n" + "=" * 80)

print("\nTARGET:")
print(target_prompt)
```

```python
from src.model import tokenize

honest_inputs = tokenize(
    bundle,
    honest_prompt,
)

target_inputs = tokenize(
    bundle,
    target_prompt,
)

print(
    "Honest input shape:",
    honest_inputs["input_ids"].shape,
)

print(
    "Target input shape:",
    target_inputs["input_ids"].shape,
)

print(
    "Token sequences identical:",
    torch.equal(
        honest_inputs["input_ids"],
        target_inputs["input_ids"],
    ),
)
```

```python
from src.patching import capture_activations

test_layers = [0, 5, 10, 15, 20, 25, 30, 35]

honest_cache = capture_activations(
    model=bundle.model,
    inputs=honest_inputs,
    layers=test_layers,
)

target_cache = capture_activations(
    model=bundle.model,
    inputs=target_inputs,
    layers=test_layers,
)

print("Activation capture complete.")
```

```python
activation_differences = {}

for layer in test_layers:

    honest_activation = (
        honest_cache.activations[layer]
    )

    target_activation = (
        target_cache.activations[layer]
    )

    if honest_activation.ndim != 3 or target_activation.ndim != 3:
        raise ValueError(
            f"Layer {layer}: expected 3D activations, got "
            f"{honest_activation.ndim}D and "
            f"{target_activation.ndim}D."
        )

    if honest_activation.shape[0] != target_activation.shape[0]:
        raise ValueError(
            f"Layer {layer}: batch-size mismatch: "
            f"{honest_activation.shape[0]} vs "
            f"{target_activation.shape[0]}"
        )

    if honest_activation.shape[2] != target_activation.shape[2]:
        raise ValueError(
            f"Layer {layer}: hidden-size mismatch: "
            f"{honest_activation.shape[2]} vs "
            f"{target_activation.shape[2]}"
        )

    # Compare the final token of each condition independently.
    honest_final = honest_activation[:, -1, :]
    target_final = target_activation[:, -1, :]

    difference = (
        honest_final - target_final
    ).abs().max().item()

    activation_differences[layer] = difference

    print(
        f"Layer {layer:2d}: {difference}"
    )
```

```python
from src.patching import run_unpatched

target_unpatched = run_unpatched(
    model=bundle.model,
    inputs=target_inputs,
)

print("Unpatched forward pass complete.")
print("Logits shape:", target_unpatched.logits.shape)
```

```python
from src.patching import run_patched

patch_results = {}

for layer in test_layers:

    patched_output = run_patched(
        model=bundle.model,
        inputs=target_inputs,
        honest_cache=honest_cache,
        layer_idx=layer,
    )

    difference = (
        target_unpatched.logits
        - patched_output.logits
    ).abs().max().item()

    patch_results[layer] = difference

    print(
        f"Layer {layer:2d}: {difference}"
    )
```

```python
repeat_results = {}

for layer in test_layers:

    patched_output = run_patched(
        model=bundle.model,
        inputs=target_inputs,
        honest_cache=honest_cache,
        layer_idx=layer,
    )

    difference = (
        target_unpatched.logits
        - patched_output.logits
    ).abs().max().item()

    repeat_results[layer] = difference

    print(
        f"Layer {layer:2d}: "
        f"first={patch_results[layer]} | "
        f"repeat={difference}"
    )
```

```python
import math

activation_pass = all(
    value > 0
    for value in activation_differences.values()
)

patch_effect_pass = all(
    value > 0
    for value in patch_results.values()
)

reproducibility_pass = all(
    math.isclose(
        patch_results[layer],
        repeat_results[layer],
        rel_tol=0.0,
        abs_tol=1e-6,
    )
    for layer in test_layers
)

positive_control_pass = (
    activation_pass
    and patch_effect_pass
    and reproducibility_pass
)

print("Activation differences:", activation_pass)
print("Patch effect:", patch_effect_pass)
print("Reproducibility:", reproducibility_pass)

print()

print(
    "POSITIVE CONTROL:",
    "PASS" if positive_control_pass else "FAIL",
)
```

```python
import json
from datetime import datetime, timezone

result = {
    "phase": "phase_2_positive_control",
    "status": (
        "PASS"
        if positive_control_pass
        else "FAIL"
    ),
    "model": config.MODEL_NAME,
    "model_revision": config.MODEL_REVISION,
    "dtype": config.DTYPE,
    "question_id": question.question_id,
    "layers_tested": test_layers,
    "activation_max_abs_difference": (
        activation_differences
    ),
    "patch_max_logit_difference": patch_results,
    "repeat_patch_max_logit_difference": (
        repeat_results
    ),
    "checks": {
        "activation_difference": activation_pass,
        "patch_effect": patch_effect_pass,
        "reproducibility": reproducibility_pass,
    },
    "repository_commit": subprocess.run(
        [
            "git",
            "-C",
            str(PROJECT_ROOT),
            "rev-parse",
            "HEAD",
        ],
        capture_output=True,
        text=True,
        check=True,
    ).stdout.strip(),
    "timestamp_utc": datetime.now(
        timezone.utc
    ).isoformat(),
}

output_path = config.POSITIVE_CONTROL_RESULT_PATH

with output_path.open(
    "w",
    encoding="utf-8",
) as file:
    json.dump(
        result,
        file,
        indent=2,
    )

print("Saved:")
print(output_path)
```
